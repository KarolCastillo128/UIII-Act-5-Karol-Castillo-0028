```python
# SIMULADOR DE INTELIGENCIA ARTIFICIAL: ASISTENTE DE ARQUITECTURA
# Basado estrictamente en condicionales y operadores lógicos

print("====================================================")
print(" 🏛️ BIENVENIDO AL SISTEMA IA DE TOMA DE DECISIONES 🏛️ ")
print("          --- CONFIGURADOR DE PROYECTOS ---         ")
print("====================================================\n")

print("El Arquitecto Senior está evaluando tu propuesta de diseño.")
print("Por favor, responde a las siguientes preguntas para que la IA tome una decisión.\n")

# --- ENTRADA DE DATOS (Interacción con el usuario) ---
tipo_proyecto = input("¿Qué tipo de proyecto es? (Residencial / Comercial / Industrial): ")
presupuesto = float(input("¿Cuál es el presupuesto estimado en USD?: "))
area_construccion = float(input("¿Cuántos metros cuadrados (m²) tiene el terreno?: "))
estilo_moderno = input("¿El cliente prefiere un estilo vanguardista/moderno? (Si / No): ")
sostenible = input("¿El proyecto requiere certificación ecológica/sustentable? (Si / No): ")

print("\n----------------------------------------------------")
print("🤖 PROCESANDO DECISIÓN DE LA IA... 🤖")
print("----------------------------------------------------\n")

# --- LÓGICA DE IA UTILIZANDO CONDICIONALES Y OPERADORES ---

# Evaluación 1: Viabilidad financiera y espacial (Cálculo del costo por metro cuadrado)
costo_por_m2 = presupuesto / area_construccion

# Decisiones anidadas y compuestas
if tipo_proyecto == "Residencial" or tipo_proyecto == "residencial":

    if costo_por_m2 < 500:
        print("❌ ALERTA DEL ARQUITECTO SENIOR: El presupuesto es demasiado bajo para un proyecto Residencial.")
        print("👉 Decisión: Rediseñar planos para reducir el área o pedir incremento de presupuesto.")
    elif costo_por_m2 >= 500 and costo_por_m2 <= 1500:
        print("✅ DICTAMEN: Presupuesto Residencial estándar adecuado.")

        # Evaluación de estilo y sostenibilidad
        if estilo_moderno == "Si" and sostenible == "Si":
            print("💎 Recomendación: Diseñar casa inteligente con paneles solares y fachada de concreto visto.")
        elif estilo_moderno == "Si" and sostenible == "No":
            print("🎨 Recomendación: Enfoque en estética minimalista, pero utilizando materiales tradicionales.")
        elif estilo_moderno == "No" and sostenible == "Si":
            print("🏡 Recomendación: Estilo rústico/colonial utilizando materiales locales y aislamiento térmico natural.")
        else:
            print("📦 Recomendación: Diseño residencial básico y funcional.")

    else:
        print("⭐ DICTAMEN: ¡Presupuesto Residencial de Lujo detectado!")
        print("👉 Decisión: Asignar al equipo de acabados premium. Se autorizan estructuras complejas y grandes ventanales.")

elif tipo_proyecto == "Comercial" or tipo_proyecto == "comercial":

    if area_construccion < 200 and presupuesto < 50000:
        print("🏪 Decisión IA: Clasificado como 'Pequeño Comercio'.")
        print("👉 Recomendación: Optimizar el espacio al máximo. Priorizar distribución fluida y zonas de cajas.")
    elif area_construccion >= 200 and presupuesto >= 50000:
        print("🏢 Decisión IA: Clasificado como 'Centro Comercial o Corporativo'.")

        if sostenible == "Si":
            print("🌿 Requisito obligatorio: El Arquitecto Senior exige incluir techos verdes y sistemas de reciclaje de agua.")
        else:
            print("⚡ Requisito obligatorio: Priorizar sistemas de iluminación eficiente y fachadas de vidrio muro cortina.")
    else:
        print("⚠️ Inconsistencia: El área y el presupuesto comercial no son proporcionales. Revisar con el cliente.")

elif tipo_proyecto == "Industrial" or tipo_proyecto == "industrial":

    # Operador lógico estricto para naves industriales
    if area_construccion > 1000 and presupuesto >= 200000:
        print("🏭 DICTAMEN: Proyecto de Gran Escala Industrial Aprobado.")
        print("👉 Decisión: Diseñar estructura de acero con grandes luces (claros libres) y pavimentos de alta resistencia.")
    else:
        print("❌ Rechazado por el Director: Las dimensiones o el presupuesto no cumplen con el mínimo para una nave industrial viable.")

else:
    print("❓ ERROR DE SINTAXIS EN EL PROYECTO: Tipo de proyecto no reconocido por la IA.")
    print("👉 Por favor, asegúrate de escribir 'Residencial', 'Comercial' o 'Industrial'.")

print("\n====================================================")
print("             FIN DE LA SIMULACIÓN DE IA             ")
print("====================================================")
print(" programa elaborado por Karol Castillo NL = 9 ")
```
